````python
# 1. ALWAYS IMPORT UNSLOTH FIRST
import unsloth
from unsloth import FastLanguageModel

import torch
from datasets import Dataset
from transformers import TrainingArguments
from trl import SFTTrainer

# =====================================================================
# 1. LOAD PRE-TRAINED BASE MODEL IN 4-BIT (QLoRA)
# =====================================================================
max_seq_length = 2048

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="Qwen/Qwen2.5-7B-Instruct",
    max_seq_length=max_seq_length,
    load_in_4bit=True,
)

# =====================================================================
# 2. ATTACH LORA ADAPTERS (r=16, alpha=32)
# =====================================================================
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    lora_alpha=32,
    lora_dropout=0,
    target_modules=[
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj"
    ],
)

# =====================================================================
# 3. PREPARE TRAINING DATA WITH CHATML
# =====================================================================
raw_training_data = [
    {
        "prompt": "Extract details from invoice: Acme Corp billed on 2024-01-15 for 2 items totaling $300.00.",
        "response": '{"vendor_name": "Acme Corp", "invoice_date": "2024-01-15", "total_amount_usd": 300.00, "items_count": 2}'
    },
    {
        "prompt": "Extract details: Nova Logistics sent a bill on 2024-03-22 for 5 shipments at $1250.75.",
        "response": '{"vendor_name": "Nova Logistics", "invoice_date": "2024-03-22", "total_amount_usd": 1250.75, "items_count": 5}'
    },
    {
        "prompt": "Invoice from Apex Systems dated 2024-06-10 with 1 service license for $99.99.",
        "response": '{"vendor_name": "Apex Systems", "invoice_date": "2024-06-10", "total_amount_usd": 99.99, "items_count": 1}'
    },
    {
        "prompt": "Parse this bill: Delta Services, invoice date 2024-08-04, 8 units purchased, balance $4500.00.",
        "response": '{"vendor_name": "Delta Services", "invoice_date": "2024-08-04", "total_amount_usd": 4500.00, "items_count": 8}'
    },
    {
        "prompt": "Extract JSON: CloudHost Inc, 2024-05-12, total cost $249.99 across 3 line items.",
        "response": '{"vendor_name": "CloudHost Inc", "invoice_date": "2024-05-12", "total_amount_usd": 249.99, "items_count": 3}'
    }
]

def apply_chatml(example):
    messages = [
        {
            "role": "system",
            "content": "You are a precise data extraction system. Output strictly a flat JSON object with keys: vendor_name, invoice_date, total_amount_usd, items_count."
        },
        {"role": "user", "content": example["prompt"]},
        {"role": "assistant", "content": example["response"]}
    ]
    return {"text": tokenizer.apply_chat_template(messages, tokenize=False)}

formatted_dataset = Dataset.from_list([apply_chatml(item) for item in raw_training_data])

# =====================================================================
# 4. CONFIGURE SFT TRAINER & TRAIN
# =====================================================================
trainer = SFTTrainer(
    model=model,
    processing_class=tokenizer,
    train_dataset=formatted_dataset,
    dataset_text_field="text",
    max_seq_length=max_seq_length,
    args=TrainingArguments(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=2,
        warmup_steps=5,
        max_steps=30,
        learning_rate=2e-4,
        fp16=not torch.cuda.is_bf16_supported(),
        bf16=torch.cuda.is_bf16_supported(),
        logging_steps=5,
        output_dir="outputs_chatml",
        report_to="none"
    ),
)

print("Starting SFT training...")
trainer.train()

model.save_pretrained("qwen_lora_model")
tokenizer.save_pretrained("qwen_lora_model")
print("LoRA SFT weights saved successfully.")
````

````python
import torch
from datasets import Dataset
from transformers import TrainingArguments
from trl import DPOTrainer
from unsloth import FastLanguageModel

# Ensure model and tokenizer are active
if "tokenizer" not in globals() or "model" not in globals():
    model, tokenizer = FastLanguageModel.from_pretrained(
        model_name="qwen_lora_model",
        max_seq_length=1024,
        load_in_4bit=True,
    )

# 1. DEFINE CONTRASTIVE PREFERENCE TRIPLETS
dpo_raw_data = [
    {
        "prompt": "Extract details from invoice: Acme Corp billed on 2024-01-15 for 2 items totaling $300.00.",
        "chosen": '{"vendor_name": "Acme Corp", "invoice_date": "2024-01-15", "total_amount_usd": 300.0, "items_count": 2}',
        "rejected": 'Sure! Here is the output:\n```json\n{"invoice": {"vendor": "Acme Corp", "date": "2024-01-15", "amount": 300.0, "items": 2}}\n```\nHope that helps!'
    },
    {
        "prompt": "Extract details: Nova Logistics sent a bill on 2024-03-22 for 5 shipments at $1250.75.",
        "chosen": '{"vendor_name": "Nova Logistics", "invoice_date": "2024-03-22", "total_amount_usd": 1250.75, "items_count": 5}',
        "rejected": '```json\n{"vendor_name": "Nova Logistics", "invoice_date": "2024-03-22", "total_amount_usd": 1250.75, "items_count": 5, "tracking_status": "delivered"}\n```'
    },
    {
        "prompt": "Invoice from Apex Systems dated 2024-06-10 with 1 service license for $99.99.",
        "chosen": '{"vendor_name": "Apex Systems", "invoice_date": "2024-06-10", "total_amount_usd": 99.99, "items_count": 1}',
        "rejected": '{"company": "Apex Systems", "date": "2024-06-10", "price": 99.99, "licenses": 1}'
    },
    {
        "prompt": "Parse this bill: Delta Services, invoice date 2024-08-04, 8 units purchased, balance $4500.00.",
        "chosen": '{"vendor_name": "Delta Services", "invoice_date": "2024-08-04", "total_amount_usd": 4500.0, "items_count": 8}',
        "rejected": 'Here is the extraction:\n{"vendor_name": "Delta Services", "invoice_date": "2024-08-04", "total_amount_usd": 4500.0}'
    }
]

# 2. FORMAT PROMPTS WITH CHATML
formatted_dpo = []
for item in dpo_raw_data:
    messages = [
        {"role": "system", "content": "You are a precise data extraction system. Output strictly a flat JSON object with keys: vendor_name, invoice_date, total_amount_usd, items_count."},
        {"role": "user", "content": item["prompt"]}
    ]
    prompt_str = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    formatted_dpo.append({
        "prompt": prompt_str,
        "chosen": item["chosen"],
        "rejected": item["rejected"]
    })

dpo_dataset = Dataset.from_list(formatted_dpo)

# 3. CONFIGURE DPO TRAINER
dpo_trainer = DPOTrainer(
    model=model,
    ref_model=None, # Unsloth computes reference outputs without an extra copy in VRAM
    processing_class=tokenizer,
    train_dataset=dpo_dataset,
    beta=0.1,
    max_length=1024,
    args=TrainingArguments(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=2,
        warmup_steps=2,
        max_steps=15,
        learning_rate=5e-6, # Reduced LR to preserve base knowledge
        fp16=not torch.cuda.is_bf16_supported(),
        bf16=torch.cuda.is_bf16_supported(),
        logging_steps=3,
        output_dir="outputs_dpo",
        report_to="none"
    )
)

print("Starting DPO training...")
dpo_trainer.train()

# 4. SAVE FINAL DPO-ALIGNED MODEL
model.save_pretrained("qwen_dpo_aligned_model")
tokenizer.save_pretrained("qwen_dpo_aligned_model")
print("\nDPO preference alignment complete and weights saved!")
````

````python
import json
from pydantic import BaseModel, ValidationError
from unsloth import FastLanguageModel

# 1. DEFINE PYDANTIC VALIDATION SCHEMA
class InvoiceExtraction(BaseModel):
    vendor_name: str
    invoice_date: str
    total_amount_usd: float
    items_count: int

# 2. UNSEEN HOLDOUT TEST SUITE
test_cases = [
    'Invoice from Stellar Cloud Services on 2025-02-14 for 12 server instances totaling $3,400.00.',
    'Bill paid to Quantum Security LLC, dated 2025-03-01. Total: $850.25 for 2 audit packages.',
    'Receipt: Horizon Data Corp | Date: 2025-04-19 | Qty: 6 | Balance Due: $1,920.00',
    'Please extract: Vertex AI Labs billed $640.50 across 3 GPU nodes on 2025-05-30.',
    'Order summary from Pinnacle Freight on 2025-06-11: 9 pallets shipped, total cost $5,100.00.'
]

FastLanguageModel.for_inference(model)

passed = 0
print("Running Post-DPO Evaluation Benchmark...\n" + "="*55)

for i, prompt_text in enumerate(test_cases, 1):
    messages = [
        {
            "role": "system",
            "content": "You are a precise data extraction system. Output strictly a flat JSON object with keys: vendor_name, invoice_date, total_amount_usd, items_count."
        },
        {"role": "user", "content": prompt_text}
    ]

    formatted_prompt = tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = tokenizer([formatted_prompt], return_tensors="pt").to("cuda")

    outputs = model.generate(
        **inputs,
        max_new_tokens=64,
        use_cache=True,
        do_sample=False, # Deterministic greedy decoding (replaces temperature=0.0)
        eos_token_id=tokenizer.encode("<|im_end|>")[0],
    )

    decoded = tokenizer.batch_decode(outputs)[0]
    raw_reply = decoded.split("<|im_start|>assistant\n")[-1].replace("<|im_end|>", "").strip()

    # Validate structure and types
    try:
        parsed = json.loads(raw_reply)
        validated = InvoiceExtraction(**parsed)
        passed += 1
        print(f"Test {i} [PASS]: {raw_reply}")
    except (json.JSONDecodeError, ValidationError) as e:
        print(f"Test {i} [FAIL]: {raw_reply} -> Error: {e}")

compliance_rate = (passed / len(test_cases)) * 100
print("="*55)
print(f"Final Schema Compliance Rate: {passed}/{len(test_cases)} ({compliance_rate:.1f}%)")
````
